## 과제. 건축연도로 매매가격을 설명하는 단순회귀 분석

`YearBuilt`를 독립변수, `SalePrice`를 종속변수로 사용하여 필수 문제와 동일한 수준의 단순선형회귀 분석을 독립적으로 수행하세요.

### 수행 요구사항

1. 최소제곱법으로 기울기와 절편을 직접 계산하세요.
2. 예측값, 잔차, SSE, SST, $R^2$을 계산하세요.
3. 건축연도가 2000년인 주택의 매매가격을 예측하세요.
4. 산점도와 회귀직선, 잔차 플롯을 그리세요.
5. 아래 질문에 답하는 5~7문장의 결론을 작성하세요.

### 질문

- 기울기와 절편은 각각 무엇을 의미하나요?
- $R^2$을 기준으로 모델의 설명력은 어떠한가요?
- 잔차 플롯을 기준으로 모델의 적합성에 주의할 점이 있나요?
- 이 모델의 적용 한계 또는 개선 방향을 한 가지 이상 제시하세요.

> 과제는 필수 문제와 동일한 수준이며, 새로운 고급 기법을 사용할 필요는 없습니다.


```python
# 여기에 코드를 작성하세요.
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. 결측치 제거 및 변수 지정
df_clean = df[["YearBuilt", "SalePrice"]].dropna()
x = df_clean["YearBuilt"].values
y = df_clean["SalePrice"].values
n = len(x)

# 1. 최소제곱법(OLS) 공식으로 기울기(beta_1)와 절편(beta_0) 직접 계산
# beta_1 = Cov(x, y) / Var(x) = Σ((x - x_mean) * (y - y_mean)) / Σ((x - x_mean)^2)
x_mean = np.mean(x)
y_mean = np.mean(y)

beta_1 = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
beta_0 = y_mean - beta_1 * x_mean

# 2. 예측값, 잔차, SSE, SST, R^2 계산
y_pred = beta_0 + beta_1 * x
residuals = y - y_pred

sse = np.sum(residuals**2)  # 잔차 제곱합
sst = np.sum((y - y_mean) ** 2)  # 총 제곱합
r_squared = 1 - (sse / sst)

# 3. 건축연도가 2000년인 주택의 매매가격 예측
pred_2000 = beta_0 + beta_1 * 2000

print("=== 단순회귀 분석 결과 ===")
print(f"기울기 (Slope, beta_1) : {beta_1:,.2f} 달러/년")
print(f"절편 (Intercept, beta_0): {beta_0:,.2f} 달러")
print(f"회귀식: SalePrice = {beta_0:,.2f} + {beta_1:,.2f} * YearBuilt")
print(f"SSE (오차 제곱합)       : {sse:,.2e}")
print(f"SST (총 제곱합)         : {sst:,.2e}")
print(f"결정계수 (R^2)          : {r_squared:.4f} ({r_squared*100:.2f}%)")
print(f"2000년 건축 주택 예측가 : ${pred_2000:,.2f}")

# 4. 산점도와 회귀직선, 잔차 플롯 시각화
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# (1) 산점도 및 회귀직선
axes[0].scatter(
    x, y, alpha=0.3, color="steelblue", edgecolors="none", s=20, label="Observed"
)
x_line = np.linspace(x.min(), x.max(), 100)
y_line = beta_0 + beta_1 * x_line
axes[0].plot(
    x_line,
    y_line,
    color="crimson",
    linewidth=2,
    label=f"Fit: y = {beta_0:,.0f} + {beta_1:,.0f}x",
)
axes[0].set_title("SalePrice vs YearBuilt with Regression Line", fontsize=11)
axes[0].set_xlabel("Year Built", fontsize=10)
axes[0].set_ylabel("SalePrice ($)", fontsize=10)
axes[0].legend()
axes[0].grid(True, linestyle=":", alpha=0.6)

# (2) 잔차 플롯 (Residual Plot)
axes[1].scatter(
    y_pred,
    residuals,
    alpha=0.3,
    color="darkorange",
    edgecolors="none",
    s=20,
)
axes[1].axhline(0, color="black", linestyle="--", linewidth=1.2)
axes[1].set_title("Residual Plot (Residuals vs Fitted Values)", fontsize=11)
axes[1].set_xlabel("Fitted Values ($)", fontsize=10)
axes[1].set_ylabel("Residuals ($)", fontsize=10)
axes[1].grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

**Q1.** 기울기와 절편은 각각 무엇을 의미하나요?
- 기울기는 건축 연도에 따른 매매 가격의 변화율을 의미하고, 절편은 건축연도가 0년일때의 매매가격으로 현실에는 존재하지 않는 수학적 값이다.

**Q2.** $R^2$을 기준으로 모델의 설명력은 어떠한가요?
- 결정계수는 약 27.34%로, 건축연도라는 지표가 매매가격의 변화를 27.34% 설명할 수 있음을 의미한다. 

**Q3.** 잔차 플롯을 기준으로 모델의 적합성에 주의할 점이 있나요?
- 예측가격이 높아질수록 잔차의 분산이 커져, 등분산성을 위반하고 있다.
