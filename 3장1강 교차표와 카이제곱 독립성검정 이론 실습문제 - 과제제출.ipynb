## 필수 1. 교차표와 카이제곱 통계량 직접 계산

### 문제 1-1. 주택 품질 구간과 중앙 냉방시설의 관계

#### 문제 설명

`OverallQual`을 세 구간으로 나눈 뒤 중앙 냉방시설 유무와의 관계를 확인합니다.

- 낮음: 1~5점
- 보통: 6~7점
- 높음: 8~10점

#### 요구사항

1. `pd.cut()`을 이용해 위 기준으로 `QualityGroup`을 만드세요.
2. 행에는 `QualityGroup`, 열에는 `CentralAir`가 오도록 합계 없는 교차표를 만드세요.
3. `margins=True`인 교차표도 별도로 만들어 행·열 합계를 확인하세요.
4. 합계가 없는 교차표를 NumPy 배열 `observed`로 변환하세요.
5. 행 합계, 열 합계, 전체 합계를 계산하세요.
6. `(행 합계 × 열 합계) / 전체 합계`로 기대빈도를 직접 계산하세요.
7. `Σ(관측-기대)²/기대`로 카이제곱 통계량을 직접 계산하세요.
8. `(행 수-1) × (열 수-1)`로 자유도를 계산하세요.
9. 모든 기대빈도가 5 이상인지 확인하세요.
10. `stats.chi2_contingency(..., correction=False)` 결과와 직접 계산한 값을 비교하세요.
11. p-value를 이용해 두 변수가 관련 있는지 판단하세요.

#### 해석 질문

**Q1.** 교차표 각 칸의 숫자는 무엇을 의미하나요?  
**Q2.** 기대빈도는 어떤 가정 아래 계산되는 값인가요?  
**Q3.** 직접 계산한 카이제곱 통계량과 함수의 결과는 일치하나요?  
**Q4.** 주택 품질 구간과 중앙 냉방시설 유무는 서로 독립이라고 볼 수 있나요?

#### 제출 결과

- 관측빈도 교차표와 주변합
- 기대빈도
- 수동 계산한 카이제곱 통계량과 자유도
- 기대빈도 조건 확인
- 함수 결과와의 비교
- 독립성 판단
- Q1~Q4 답변

```python
# 필수 1 코드를 작성하세요.
import numpy as np
import pandas as pd
from scipy import stats

# 1. pd.cut()을 이용한 QualityGroup 구간 생성 (1~5: 낮음, 6~7: 보통, 8~10: 높음)
bins = [0, 5, 7, 10]
labels = ["낮음", "보통", "높음"]
df["QualityGroup"] = pd.cut(df["OverallQual"], bins=bins, labels=labels)

# 2. 합계 없는 관측빈도 교차표
ct = pd.crosstab(df["QualityGroup"], df["CentralAir"])
print("=== 1. 관측빈도 교차표 ===")
print(ct)

# 3. margins=True 교차표 (주변합 확인)
ct_margins = pd.crosstab(
    df["QualityGroup"], df["CentralAir"], margins=True, margins_name="합계"
)
print("\n=== 2. 주변합 포함 교차표 ===")
print(ct_margins)

# 4. NumPy 배열로 변환
observed = ct.values

# 5. 행 합계, 열 합계, 전체 합계 계산
row_totals = observed.sum(axis=1, keepdims=True)  # (3, 1)
col_totals = observed.sum(axis=0, keepdims=True)  # (1, 2)
total = observed.sum()

# 6. 기대빈도 직접 계산: (행 합계 * 열 합계) / 전체 합계
expected = (row_totals @ col_totals) / total
expected_df = pd.DataFrame(
    expected, index=ct.index, columns=ct.columns
)
print("\n=== 3. 직접 계산한 기대빈도 ===")
print(expected_df.round(2))

# 7. 카이제곱 통계량 직접 계산: Σ(관측 - 기대)² / 기대
chi2_manual = np.sum(((observed - expected) ** 2) / expected)

# 8. 자유도 계산: (행 수 - 1) * (열 수 - 1)
r, c = observed.shape
df_manual = (r - 1) * (c - 1)

# p-value 직접 계산 (카이제곱 분포 생존함수)
p_manual = stats.chi2.sf(chi2_manual, df_manual)

# 9. 기대빈도 5 이상 확인
min_expected = expected.min()
all_above_5 = bool(min_expected >= 5)

print("\n=== 4. 수동 계산 결과 ===")
print(f"카이제곱 통계량 (수동) : {chi2_manual:.4f}")
print(f"자유도 (수동)          : {df_manual}")
print(f"p-value (수동)         : {p_manual:.4e}")
print(f"최소 기대빈도          : {min_expected:.2f} (5 이상 충족: {all_above_5})")

# 10. scipy.stats.chi2_contingency() 결과 대조 (correction=False)
chi2_stat, p_val, dof, expected_func = stats.chi2_contingency(
    observed, correction=False
)

print("\n=== 5. scipy 함수 결과 대조 ===")
print(f"함수 반환 카이제곱 통계량: {chi2_stat:.4f}")
print(f"함수 반환 자유도         : {dof}")
print(f"함수 반환 p-value        : {p_val:.4e}")
print(f"통계량 일치 여부         : {np.isclose(chi2_manual, chi2_stat)}")
print(f"기대빈도 일치 여부       : {np.allclose(expected, expected_func)}")

# 11. 독립성 판정
alpha = 0.05
print("\n=== 6. 독립성 판정 ===")
if p_val < alpha:
    print(
        f"판정: p-value({p_val:.4e}) < {alpha} 이므로 두 변수는 독립이 아니며 관련이 있습니다."
    )
else:
    print(
        f"판정: p-value({p_val:.4e}) >= {alpha} 이므로 두 변수는 서로 독립입니다."
    )
```

#### 제출 결과

- 교차표와 가설
- 카이제곱 검정 결과
- 기대빈도 조건
- 행 비율
- 관련성 및 인과관계 해석
- Q1~Q5 답변

```python
# 필수 2 코드를 작성하세요.
import numpy as np
import pandas as pd
from scipy import stats

# 1. pd.cut()으로 YearBuiltGroup 구간 생성
# 구간: 1980년 이전 (최소연도~1979), 1980~1999년, 2000년 이후 (2000~최대연도)
bins = [0, 1979, 1999, df["YearBuilt"].max() + 1]
labels = ["1980년 이전", "1980~1999년", "2000년 이후"]

df["YearBuiltGroup"] = pd.cut(
    df["YearBuilt"], bins=bins, right=False, labels=labels
)

# 2. YearBuiltGroup과 CentralAir 교차표 생성
ct = pd.crosstab(df["YearBuiltGroup"], df["CentralAir"])
print("=== 1. 교차표 (관측빈도) ===")
print(ct)

# 4 & 5. 카이제곱 독립성 검정 수행 (stats.chi2_contingency)
chi2_stat, p_val, dof, expected = stats.chi2_contingency(
    ct.values, correction=False
)
expected_df = pd.DataFrame(expected, index=ct.index, columns=ct.columns)

print("\n=== 2. 기대빈도 표 ===")
print(expected_df.round(2))

print("\n=== 3. 카이제곱 검정 통계치 ===")
print(f"카이제곱 통계량 : {chi2_stat:.4f}")
print(f"자유도 (df)     : {dof}")
print(f"p-value         : {p_val:.4e}")

# 6. 기대빈도 중 5 미만인 칸 수 확인
under_5_count = (expected < 5).sum()
print(f"\n기대빈도 5 미만 칸 수: {under_5_count}개")

# 7. 행 기준 정규화 비율 (구간별 냉방시설 비율)
row_ratio = pd.crosstab(
    df["YearBuiltGroup"], df["CentralAir"], normalize="index"
)
print("\n=== 4. 행 비율 (YearBuiltGroup별 CentralAir 비율) ===")
print((row_ratio * 100).round(2).astype(str) + "%")

# 8. 판정
alpha = 0.05
print("\n=== 5. 판정 ===")
if p_val < alpha:
    print(
        f"p-value({p_val:.4e}) < {alpha} 이므로 귀무가설(H₀)을 기각합니다 (통계적으로 유의한 관련성 있음)."
    )
else:
    print(
        f"p-value({p_val:.4e}) >= {alpha} 이므로 귀무가설(H₀)을 기각하지 못합니다."
    )
```

**Q4.** 두 주방 품질 집단의 완전 포장 비율은 각각 얼마인가요?

#### 제출 결과

- 재분류 코드와 교차표
- 가설 설정
- 카이제곱 검정 결과
- 기대빈도 조건 확인
- 행 비율과 결과 해석
- Q1~Q4 답변

```python
# 과제 코드를 작성하세요.
import numpy as np
import pandas as pd
from scipy import stats

# 1. 두 변수 결측치 제외 및 재분류(KitchenGroup, DriveGroup)
df_clean = df[["KitchenQual", "PavedDrive"]].dropna().copy()

kitchen_map = {"Ex": "우수", "Gd": "우수", "TA": "보통 이하", "Fa": "보통 이하"}
drive_map = {"Y": "완전 포장", "N": "미포장·부분포장", "P": "미포장·부분포장"}

df_clean["KitchenGroup"] = df_clean["KitchenQual"].map(kitchen_map)
df_clean["DriveGroup"] = df_clean["PavedDrive"].map(drive_map)

# 2. 교차표 작성 (주변합 포함 및 미포함)
crosstab_raw = pd.crosstab(
    df_clean["KitchenGroup"], df_clean["DriveGroup"]
)
crosstab_margins = pd.crosstab(
    df_clean["KitchenGroup"], df_clean["DriveGroup"], margins=True
)

# 4. 카이제곱 독립성 검정 수행 (correction=False: 수동 통계량과 일관성 유지)
chi2_stat, p_val, dof, expected = stats.chi2_contingency(
    crosstab_raw.values, correction=False
)
expected_df = pd.DataFrame(
    expected, index=crosstab_raw.index, columns=crosstab_raw.columns
)

# 6. 기대빈도 조건 점검 (모든 셀 >= 5)
all_ge_5 = np.all(expected >= 5)

# 7. 주방 품질 집단별 진입로 포장 비율 (행 비율, Row Proportion)
row_props = pd.crosstab(
    df_clean["KitchenGroup"], df_clean["DriveGroup"], normalize="index"
)

# 출력
print("=== 1. 관측 교차표 (주변합 포함) ===")
print(crosstab_margins)

print("\n=== 2. 기대빈도 ===")
print(expected_df.round(2))
print(f"모든 기대빈도 >= 5 충족 여부: {all_ge_5} (최소 기대빈도: {expected.min():.2f})")

print("\n=== 3. 카이제곱 검정 결과 ===")
print(f"카이제곱 통계량 : {chi2_stat:.4f}")
print(f"자유도 (df)     : {dof}")
print(f"p-value         : {p_val:.4e}")

print("\n=== 4. 주방 품질 집단별 포장 상태 비율 (행 비율) ===")
print((row_props * 100).round(2).astype(str) + "%")
```
