## 과제. Cookie Cats A/B 테스트 전체 계획서 작성

`gate_30`을 기존 버전, `gate_40`을 새 버전으로 설정한 A/B 테스트 계획을 작성하세요.

### 수행 요구사항

1. 아래 네 단계를 모두 포함한 계획서를 작성하세요.
   - 가설 설정
   - 실험 설계
   - 실험 실행
   - 결과 분석
2. 실험 단위, 대조군, 실험군, 핵심·보조·가드레일 지표를 명시하세요.
3. 실행 전에 확인할 데이터 품질 항목을 두 가지 이상 작성하세요.
4. SUTVA 위반 또는 실험 간 간섭 가능성을 검토하세요.
5. 버전별 관측 지표를 다시 계산하되, 아직 통계적 검정을 하지 않았다는 점을 반영해 최종 의사결정을 보류하는 6~8문장의 결론을 작성하세요.

> 과제는 필수 문제와 동일한 수준입니다. 표본 크기나 MDE를 계산할 필요는 없습니다.


**가설 설정**
   - 귀무 가설 : 첫 번째 보스 조우 위치를 스테이지30에서 40으로 이동해도 retention_7에는 차이가 없다.
   - 대립 가설 : 첫 번째 보스 조우 위치를 스테이지30에서 40으로 이동하면 retention_7에는 유의미한 차이가 있을것이다.

**실험 설계**   
   - 실험단위: userid
   - 대조군gate_30, 실험군gate_40
   - 배정 비율 50:50 
   - 핵심지표 retention_7, 보조지표 retention_1 / sum_gamerounds , 가드레일지표: 1인당 인앱 결제금액, 클라이언트 크래시율 등

**실험 실행**   
   - 실험 실행 : SRM (Sample Ratio Mismatch) 검증, 극단치 및 이상치 필터링 기준 수립  
   
**결과 분석** 
   - Cookie Cats 데이터셋의 기초 통계량 집계 결과, 7일 차 유지율은 기존 버전(gate_30)의 19.02%에 비해 신규 버전(gate_40)에서 18.20%로 약 0.82%p 낮게 관측되었다. 
   - 보조 지표인 1일 차 유지율 역시 gate_30의 44.82%에서 gate_40의 44.23%로 0.59%p 소폭 하락했다. 
   - 유저 1인당 평균 플레이 판수 또한 기존 게이트 버전이 약 1.16판 더 많이 플레이한 것으로 나타났다.
   - 현재 단계의 결과는 단순 집계치일 뿐이며, 이 차이가 표본 추출 오차나 우연에 의한 노이즈인지 확인하는 통계적 가설 검정을 아직 거치지 않았다.
   - 따라서 관측치 하락만으로 gate 위치 변경을 확정해서는 안 되며, 정식 통계 검정과 신뢰구간 분석이 완료될 때까지 최종 의사결정을 보류한다.

```python
# 여기에 코드를 작성하세요.
import numpy as np
import pandas as pd

# 1. 버전별 핵심 관측 지표 집계 (표본 수, 1일 유지율, 7일 유지율, 평균 플레이 수)
summary_df = (
    df.groupby("version")
    .agg(
        user_count=("userid", "count"),
        retention_1_mean=("retention_1", "mean"),
        retention_7_mean=("retention_7", "mean"),
        gamerounds_mean=("sum_gamerounds", "mean"),
        gamerounds_median=("sum_gamerounds", "median"),
    )
    .reset_index()
)

# 2. 비율 지표 퍼센트(%p) 단위 변환 및 차이(Treatment - Control) 계산
g30 = summary_df[summary_df["version"] == "gate_30"].iloc[0]
g40 = summary_df[summary_df["version"] == "gate_40"].iloc[0]

diff_row = {
    "version": "차이 (gate_40 - gate_30)",
    "user_count": g40["user_count"] - g30["user_count"],
    "retention_1_mean": g40["retention_1_mean"] - g30["retention_1_mean"],
    "retention_7_mean": g40["retention_7_mean"] - g30["retention_7_mean"],
    "gamerounds_mean": g40["gamerounds_mean"] - g30["gamerounds_mean"],
    "gamerounds_median": g40["gamerounds_median"] - g30["gamerounds_median"],
}

# 3. 결과 테이블 포맷팅
display_df = pd.concat(
    [summary_df, pd.DataFrame([diff_row])], ignore_index=True
)

formatted_df = pd.DataFrame(
    {
        "버전": display_df["version"],
        "유저 수 (N)": display_df["user_count"].apply(
            lambda x: f"{x:+,}명" if isinstance(x, (int, np.integer)) and x < 0 else f"{x:,}명"
        ),
        "1일 유지율 (retention_1)": display_df["retention_1_mean"].apply(
            lambda x: f"{x*100:+.2f}%p" if abs(x) < 0.1 else f"{x*100:.2f}%"
        ),
        "7일 유지율 (retention_7)": display_df["retention_7_mean"].apply(
            lambda x: f"{x*100:+.2f}%p" if abs(x) < 0.1 else f"{x*100:.2f}%"
        ),
        "평균 플레이 수 (rounds)": display_df["gamerounds_mean"].apply(
            lambda x: f"{x:+.2f}판" if abs(x) < 10 else f"{x:.2f}판"
        ),
        "중앙값 플레이 수 (rounds)": display_df["gamerounds_median"].apply(
            lambda x: f"{x:+.1f}판" if abs(x) < 5 else f"{x:.1f}판"
        ),
    }
)

print("=== Cookie Cats 버전별 관측 지표 비교 (순수 집계치) ===")
print(formatted_df.to_string(index=False))
```
